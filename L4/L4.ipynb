variance

```python
import numpy as np
import statistics as st

scores = [5, 6, 7, 8, 100]

print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
print("Variance:", np.var(scores))
print("Standard deviation:", np.std(scores))
print("Sum of squared differences:", sum((np.array(scores) - np.mean(scores)) ** 2))
```

activity - student height statistics

```python

import numpy as np

height = np.array([148, 154, 105, 108, 111, 116, 133, 149])

print("Minimum:", np.min(height))
print("Maximum:", np.max(height))
print("Range:", np.max(height) - np.min(height))

print("90th percentile:", np.percentile(height, 90))

print("Mean:", np.mean(height))
print("Median:", np.median(height))

print("Variance:", np.var(height))
print("Standard deviation:", np.std(height))
```
